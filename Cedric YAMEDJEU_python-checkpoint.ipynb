## Question 4
- Interesect the rivers and the countries
- Create a chloropleth Map representing the total length of the rivers in each country

```python
import matplotlib.pyplot as plt
import pandas as pd
import geopandas as gpd 

# load shapefiles
country = gpd.read_file(r"C:\Users\bruno\Downloads\PS03\PS03\shp\countries\World_Countries.shp")
rivers = gpd.read_file(r"C:\Users\bruno\Downloads\PS03\PS03\shp\rivers\World_Hydrography.shp")

country_proj = country.to_crs("ESRI:54030")
rivers_proj = rivers.to_crs("ESRI:54030")

# Intersection country - rivers
intersection = gpd.overlay(country_proj, rivers_proj, how="intersection")

# Length calculation (in meters, converted to km)
intersection["length_km"] = intersection.geometry.length / 1000

# Aggregation by country
river_length_by_country = (
    intersection.groupby("COUNTRY")["length_km"].sum().reset_index()
)

# Join to the world's GeoDataFrame
country_rivers = country_proj.merge(river_length_by_country, on="COUNTRY", how="left")
country_rivers["length_km"] = country_rivers["length_km"].fillna(0)

# Choropleth map
fig, ax = plt.subplots(1, 1, figsize=(15, 10))
country_rivers.plot(
    column="length_km",
    cmap="Blues",
    linewidth=0.8,
    edgecolor="0.2",
    legend=True,
    ax=ax,
    legend_kwds={"label": "Total river length (km)", "orientation": "horizontal"}
)

ax.set_title("Total length of rivers by country", fontsize=20)
ax.axis("off")
plt.show()
```

## Question 6 - API 
 - Write an API request for the total population in 2019 by State and store the data in a dataframe  
Hint: API Requests can be sent with 'requests.get()'  
Hint: Decode the text returned with the package 'json' and its function 'loads()'


```python
import requests
import json
import pandas as pd
# API Request Construction 
# Census Bureau API - Population Estimates Program (PEP) for 2019
url = "https://api.census.gov/data/2019/acs/acs1"
params = {
    "get": "B01003_001E,NAME",   # NAME = name of the state, B01003_001E = total of population
    "for": "state:*",    # All state
    "key": "207a6d53ddf8971b3cad2d9656d777f87a649d96"
}

# Sending the request 
response = requests.get(url, params=params)


# Checking status
if response.status_code == 200:
    
    data = json.loads(response.text)

    # Building the DataFrame
    population_df = pd.DataFrame(data[1:], columns=data[0])

    #  Type conversion
    population_df["B01003_001E"] = population_df["B01003_001E"].astype(int)

    print(population_df.head())
else:
    print(f"Error during the request : {response.status_code}")
    print(response.text)
```

## Question 7: API 
 - Create a program to recover the population from 2011 with the API and create a panel dataset (i.e a dataset with an observation corresponding to a state for a given year)

```python
import requests
import json
import pandas as pd

API_KEY = "207a6d53ddf8971b3cad2d9656d777f87a649d96"

# Population 2011 
def get_population_2011():
    url = "https://api.census.gov/data/2014/pep/natstprc"
    params = {
        "get": "STNAME,POP", # pop: total of population
        "DATE_": 4,
        "for": "state:*",
        "key": API_KEY
    }
    response = requests.get(url, params=params)
    if response.status_code == 200:
        data = json.loads(response.text)
        df = pd.DataFrame(data[1:], columns=data[0])
        df["POP"] = df["POP"].astype(int)
        df["YEAR"] = 2011
        return df[["state", "STNAME", "YEAR", "POP"]]
    else:
        print(f"Error 2011 : {response.status_code} - {response.text}")
        return None

# Population 2019 
def get_population_2019():
    url = "https://api.census.gov/data/2019/pep/population"
    params = {
        "get": "POP,NAME",
        "for": "state:*",
        "key": API_KEY
    }
    response = requests.get(url, params=params)
    if response.status_code == 200:
        data = json.loads(response.text)
        df = pd.DataFrame(data[1:], columns=data[0])
        df = df.rename(columns={"NAME": "STNAME"})
        df["POP"] = df["POP"].astype(int)
        df["YEAR"] = 2019
        return df[["state", "STNAME", "YEAR", "POP"]]
    else:
        print(f"Error 2019 : {response.status_code} - {response.text}")
        return None

# Building the panel dataset
df_2011 = get_population_2011()
df_2019 = get_population_2019()

panel_df = pd.concat([df_2011, df_2019], ignore_index=True)
panel_df = panel_df.sort_values(["state", "YEAR"]).reset_index(drop=True)

print(panel_df.head(10))
print(f"\nPanel dimensions : {panel_df.shape}")
print(f"Present years : {sorted(panel_df['YEAR'].unique())}")
```

## Question 8: API and GIS
 - Using python, download and open the ZIP file containing the shapefile of the US States and plot it:  
 https://www.census.gov/geographies/mapping-files/time-series/geo/cartographic-boundary.html
 Hint to download a file, you can use 
 Open an url :

         file.write(res.content)  
This will simply download and  save the file  in your working directory !   
You can navigate inside zipfiles and extract its content with the library 'zipfile' !  
     
 

```python
import requests
import zipfile
import os
import geopandas as gpd
import matplotlib.pyplot as plt

# Downloading the ZIP file
url = "https://www2.census.gov/geo/tiger/GENZ2022/shp/cb_2022_us_state_20m.zip"
res = requests.get(url)

with open("us_states.zip", "wb") as file:
    file.write(res.content)

# Extracting ZIP content
extract_dir = "us_states_shapefile"
os.makedirs(extract_dir, exist_ok=True)

with zipfile.ZipFile("us_states.zip", "r") as zip_ref:
    zip_ref.extractall(extract_dir)
    print("Extracted files :", zip_ref.namelist())

# Reading the shapefile with geopandas
# The .shp file is automatically searched for in the extracted folder
shp_file = [f for f in os.listdir(extract_dir) if f.endswith(".shp")][0]
us_states = gpd.read_file(os.path.join(extract_dir, shp_file))

#print(us_states.head())
#print(f"CRS : {us_states.crs}")

# Map display
fig, ax = plt.subplots(figsize=(16, 9), facecolor='white')
ax.set_facecolor('white')

us_states.plot(ax=ax, color="lightgrey", edgecolor="black")

ax.set_title("US States", fontsize=16, color="black")
#ax.set_axis_off()

plt.show()
```

## Question 9: Merge of shape files and datasets
 - Merge the map and the dataset with the population in 2019
 - Compute the density of population per square kilometers in 2019
 - plot the density  with a chloropleth Map
 - plot another indicating states above and below the median density (don't forget the legend)

```python
import requests
import json
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

API_KEY = "207a6d53ddf8971b3cad2d9656d777f87a649d96"

# 2019 population recovery
url = "https://api.census.gov/data/2019/pep/population"
params = {"get": "POP,NAME", "for": "state:*", "key": API_KEY}
response = requests.get(url, params=params)
data = json.loads(response.text)
population_2019 = pd.DataFrame(data[1:], columns=data[0])
population_2019["POP"] = population_2019["POP"].astype(int)

# Map + population merge 
# Check the exact name of the States column in us_states (often 'NAME')
print(us_states.columns)

us_merged = us_states.merge(population_2019, on="NAME", how="left")

# Check: Unmerged states (e.g., territories without demographic data)
missing = us_merged[us_merged["POP"].isna()]
if not missing.empty:
    print("States with no associated population :", missing["NAME"].tolist())

# Density calculation (inhabitants / km²)
# Reprojection to a projected CRS for correct area calculation (meters)
# Albers Equal Area USGS, standard for the USA
us_merged_proj = us_merged.to_crs("ESRI:102003")

us_merged_proj["area_km2"] = us_merged_proj.geometry.area / 1_000_000  # m² -> km²
us_merged_proj["density"] = us_merged_proj["POP"] / us_merged_proj["area_km2"]

# Density choropleth
continental = us_merged_proj.copy()

fig, ax = plt.subplots(figsize=(16, 9), facecolor='white')
ax.set_facecolor('white')

continental.plot(
    column="density",
    ax=ax,
    cmap="OrRd",
    edgecolor="black",
    linewidth=0.3,
    legend=True,
    legend_kwds={
        "label": "Population density (inhab./km²)",
        "orientation": "horizontal",
        "shrink": 0.6,
    },
)

ax.set_title("Population Density by State (2019)", fontsize=16, color="black")
#ax.set_axis_off()

cbar = ax.get_figure().axes[-1]
cbar.xaxis.label.set_color("white")
cbar.tick_params(colors="white")

plt.show()

# Map above / below the median 
median_density = continental["density"].median()
continental["above_median"] = continental["density"] > median_density

fig2, ax2 = plt.subplots(figsize=(16, 9), facecolor='white')
ax2.set_facecolor('white')

continental.plot(
    column="above_median",
    ax=ax2,
    cmap="coolwarm",
    edgecolor="black",
    linewidth=0.3,
    categorical=True,
    legend=False, 
)

ax2.set_title(f"States Above/Below Median Density ({median_density:.1f} inhab./km²)",
              fontsize=16, color="black")
#ax2.set_axis_off()

# Manual legend
legend_elements = [
    Patch(facecolor=plt.cm.coolwarm(0.9), edgecolor="black", label="Above median"),
    Patch(facecolor=plt.cm.coolwarm(0.1), edgecolor="black", label="Below median"),
]
legend = ax2.legend(handles=legend_elements, loc="lower left",
                     facecolor="black", edgecolor="white")
for text in legend.get_texts():
    text.set_color("white")

plt.show()
```
