```python
import pandas as pd
```

```python
df = pd.read_csv("netflix_titles.csv")
```

```python
# Récupérer les trois pays qui ont le plus de films présents sur Netflix
df.country.value_counts()[:3].to_dict()
```

```python
# Récupérer le nombre de films produits chaque années
df.release_year.value_counts().sort_index(ascending=False)[:10].to_dict()
```

```python
# Récupérer la liste de tous les réalisateurs présents sur la plateforme Netflix
df["director"].dropna().unique().tolist()[:10]
```

```python
# Récupérer la durée moyenne des films présents sur Netflix
df_duration = df[df["duration"].str.contains("min")]["duration"]
df_duration.str.replace(" min", "").map(int).mean()
```
